## How it works

The benchmark has four tasks on a road network. Each task got its own model, and the outputs were merged into one submission.

- Traffic state. Gradient boosted trees fill in missing speed and density cells. The inputs mix space, time and traffic flow physics.
- Queue forecasting. A two branch forecaster predicts where queues form and grow. It blends a physics view of shockwaves with gradient boosted trees.
- Physics consistency. Predictions follow the classic flow and density laws of traffic.
- Demand estimation. An exact nonnegative least squares solver recovers origin to destination flows from link counts.
